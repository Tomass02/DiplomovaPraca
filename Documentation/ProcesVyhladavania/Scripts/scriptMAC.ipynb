```sql
/****** Object:  UserDefinedFunction [dbo].[fKlientsByEmailContact]    Script Date: 30. 10. 2026 11:54:05 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

-- =============================================
-- Author:		<Peter Révay>
-- Create date: <10.11.2015>
-- Description:	<Vrati zoznam klient_id, cnum, email kde sa vyskytuje email >
-- =============================================
CREATE FUNCTION [dbo].[fKlientsByEmailContact] 
(
	-- Add the parameters for the function here
	@email varchar(100)
)
RETURNS 
@tZoznamKlientov TABLE 
(
	-- Add the column definitions for the TABLE variable here
	klient_id varchar(10),
	cnum varchar(6),
	acnum varchar(10),
	clntsn varchar(10),
	email_kontakt varchar (50),
	zdroj varchar(50)
)
AS
BEGIN
	
	--MAC
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT klient_id,cnum,'','',kontakt,'[MAC].[dbo].[KLIENT_KONTAKT]' FROM [MAC].[dbo].[KLIENT_KONTAKT] WHERE kontakt = @email
	
	-- PROFILE DATA
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT  c.klient_id, b.CNUM, '', a.CLNTSN,a.email,'[BANKA].[dbo].[PROFILE_DATA]' FROM [BANKA].[dbo].[PROFILE_DATA] a 
	INNER JOIN Banka..ibsclnt b on a.clntsn = b.CLNTSN collate Slovak_CI_AS
	JOIN MAC..klient c on b.CNUM = c.cnum collate SQL_Latin1_General_CP1_CI_AS
	WHERE email = @email

	-- ABO
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT c.klient_id, b.CNUM, '', a.CLNTSN,a.email,'[BANKA].[dbo].[ABO_DAILY_SEND]' FROM BANKA.dbo.ABO_DAILY_SEND a 
	JOIN cop..UCET b on b.cislo_uctu collate Slovak_CI_AS = a.ACNOC 
	JOIN MAC..klient c on b.CNUM = c.cnum collate Slovak_CI_AI
	WHERE email = @email

	-- VYPIS
	INSERT INTO @tZoznamKlientov
	SELECT DISTINCT b.klient_id,b.cnum,b.cislo_uctu,'',a.EMAIL,'[VYPISY].[dbo].[VYPISY_ELEKTRONIC]'
	FROM VYPISY.dbo.VYPISY_ELEKTRONIC a
	join COP.dbo.UCET b on b.cislo_uctu =a.ACCNOC
	WHERE email = @email

	--denne pohyby
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT  c.klient_id, b.CNUM, a.ACNOC, a.CLNTSN,a.email,'[BANKA].[dbo].[EMAIL_DAILY_AMOUNTS]' FROM [BANKA].[dbo].[EMAIL_DAILY_AMOUNTS] a 
	JOIN cop..UCET b on b.cislo_uctu collate Slovak_CI_AS = a.ACNOC 
	JOIN MAC..klient c on b.CNUM  = c.cnum 
	WHERE email =@email

	--karta 
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT  c.klient_id, b.CNUM, a.ACNOC, a.name,a.email,'[BANKA].[dbo].[EMAIL_H2H]' FROM [BANKA].[dbo].[EMAIL_H2H] a 
	JOIN cop..UCET b on b.cislo_uctu collate SQL_Latin1_General_CP1_CI_AS = a.ACNOC
	JOIN MAC..klient c on b.CNUM  = c.cnum 
	WHERE email=@email

	-- kreditne obraty
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT  c.klient_id, b.CNUM, a.ACNOC, a.[CLNTSN],a.email,'[BANKA].[dbo].[EMAIL_KREDIT]' FROM [BANKA].[dbo].[EMAIL_KREDIT] a 
	JOIN cop..UCET b on b.cislo_uctu collate SQL_Latin1_General_CP1_CI_AS = a.ACNOC 
	JOIN MAC..klient c on b.CNUM = c.cnum 
	WHERE email =@email

	-- debetne obraty
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT  c.klient_id, b.CNUM, a.ACNOC, a.[CLNTSN],a.email,'[BANKA].[dbo].[EMAIL_DEBET]' FROM [BANKA].[dbo].[EMAIL_DEBET] a 
	JOIN cop..UCET b on b.cislo_uctu collate SQL_Latin1_General_CP1_CI_AS= a.ACNOC 
	JOIN MAC..klient c on b.cnum = c.cnum
	WHERE email =@email

	--vypis dennych pohybov
	INSERT INTO @tZoznamKlientov 
  	SELECT DISTINCT  c.klient_id, b.CNUM, a.ACNOC, a.[CLNTSN],a.email,'[BANKA].[dbo].[MOVMENTS_DAILY_SEND]' FROM [BANKA].[dbo].[MOVMENTS_DAILY_SEND] a 
	JOIN cop..UCET b on b.cislo_uctu collate Slovak_CI_AS = a.ACNOC 
	JOIN MAC..klient c on b.cnum = c.cnum
	WHERE email=@email		

	-- vypis
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT klient_id,cnum,cislo_uctu,'',email,'[COP].[dbo].[UCET_ZIADOST_ZMENA_VYPISOV]' from [COP].[dbo].[UCET_ZIADOST_ZMENA_VYPISOV]
	WHERE email =@email and sposob_zasielania ='EL'

	-- vypis
	INSERT INTO @tZoznamKlientov 
	SELECT DISTINCT klient_id,cnum,ucet_cislo_uctu,'',ucet_vypis_email,'[COP].[dbo].[UCET_ZIADOST_ZMENA_VYPISOV]' from [COP].[dbo].UCET_ZIADOST
	WHERE ucet_vypis_email =@email

	RETURN 	
END
GO
```

# [dbo].[fKlietsByPhoneContact]

```sql
/****** Object:  UserDefinedFunction [dbo].[fKlietsByPhoneContact]    Script Date: 30. 10. 2026 11:54:05 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

-- =============================================
-- Author:		<Majercak,,Jan>
-- Create date: <20.4.2015>
-- Description:	<Vrati zoznam klient_id, cnum, tel_cislo cez cast. tel. cisla porovnava poslenych 9-cisel>
-- =============================================
CREATE FUNCTION [dbo].[fKlietsByPhoneContact] 
(
	-- Add the parameters for the function here
	@ptelc varchar(20)
)
RETURNS 
@tZoznamKlientov TABLE 
(
	-- Add the column definitions for the TABLE variable here
	klient_id varchar(10),
	cnum varchar(6),
	acnum varchar(10),
	clntsn varchar(10),
	m_kontakt varchar (20),
	zdroj varchar(40)
)


AS
BEGIN
declare @ptelc_e varchar(20);
declare  @tbl_zk TABLE (klient_id varchar(10),
	cnum varchar(6),
	acnum varchar(10),
	clntsn varchar(10),
	m_kontakt varchar (20),
	zdroj varchar(40)
	)
	--kontrola tel cisla
	set @ptelc = LTRim(RTRIM(@ptelc))
	set @ptelc = replace(@ptelc,' ','')
	set @ptelc = replace(@ptelc,'/','')
	set @ptelc = replace(@ptelc,'-','')
	set @ptelc_e = @ptelc
	set @ptelc = right(@ptelc,9)
	if (left(@ptelc_e,1) = '0') Set @ptelc_e = Right(@ptelc_e, Len(@ptelc_e)-1)
	if (left(@ptelc_e,4) = '+421') Set @ptelc_e = Right(@ptelc_e, Len(@ptelc_e)-4)

	--MAC
	insert into @tbl_zk select klient_id,cnum,'','',kontakt,'[MAC].[dbo].[KLIENT_KONTAKT]' from [MAC].[dbo].[KLIENT_KONTAKT] where Right(RTRIM(kontakt),9) = @ptelc 
	insert into @tbl_zk select klient_id,cnum,'','',kontakt,'[MAC].[dbo].[KLIENT_KONTAKT]' from [MAC].[dbo].[KLIENT_KONTAKT] where  (Right(RTRIM(kontakt),Len(@ptelc_e)) = @ptelc_e)

	--UVEX
	insert into @tbl_zk select klient_id,Left(Ltrim(RTrim(cnum)),6),'','',LEFT(LTRIM(RTRIM(kontakt)),20),'[UVEX].[dbo].[ZIADOST_KLIENT_KONTAKT]' from [UVEX].[dbo].[ZIADOST_KLIENT_KONTAKT] where Right(RTRIM(kontakt),9) = @ptelc
	insert into @tbl_zk select klient_id,Left(Ltrim(RTrim(cnum)),6),'','',LEFT(LTRIM(RTRIM(kontakt)),20),'[UVEX].[dbo].[ZIADOST_KLIENT_KONTAKT]' from [UVEX].[dbo].[ZIADOST_KLIENT_KONTAKT] where (Right(RTRIM(kontakt),Len(@ptelc_e)) = @ptelc_e)

	--z e-banky SMS_KREDIT
	insert into @tZoznamKlientov select c.klient_id, b.cnum, a.ACNOC, a.CLNTSN,a.MOBILE,'[BANKA].[dbo].[SMS_KREDIT]' from [BANKA].[dbo].[SMS_KREDIT] a 
	inner join [COP].[dbo].[UCET] b on a.acnoc = b.cislo_uctu collate Slovak_CI_AI
	inner join MAC..klient c on b.cnum = c.cnum collate Slovak_CI_AI
	where Right(RTRIM(a.MOBILE),9) = right(@ptelc,9)
	
	--z e-banky SMS_DEBET
	insert into @tbl_zk select c.klient_id, b.cnum, a.ACNOC, a.CLNTSN,a.MOBILE,'[BANKA].[dbo].[KLIENT_DEBET]' from [BANKA].[dbo].[SMS_DEBET] a 
	inner join [COP].[dbo].[UCET] b on a.acnoc = b.cislo_uctu collate Slovak_CI_AI
	inner join MAC..klient c on b.cnum = c.cnum collate Slovak_CI_AI
	where Right(RTRIM(a.MOBILE),9) = right(@ptelc,9)
	--z e-banky SMS_H2H
	insert into @tbl_zk select c.klient_id, b.cnum, a.ACNOC, a.CLNTSN,a.MOBILE,'[BANKA].[dbo].[KLIENT_H2H]' from [BANKA].[dbo].[SMS_H2H] a 
	inner join [COP].[dbo].[UCET] b on a.acnoc = b.cislo_uctu collate Slovak_CI_AI
	join MAC..klient c on b.cnum = c.cnum collate Slovak_CI_AI
	where Right(RTRIM(a.MOBILE),9) = right(@ptelc,9)
	--z e-banky profile data
	insert into @tbl_zk select c.klient_id, b.CNUM, '', a.CLNTSN,a.mobil,'[BANKA].[dbo].[PROFILE_DATA]' from [BANKA].[dbo].[PROFILE_DATA] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where Right(RTrim(a.mobil),9) = @ptelc

	insert into @tbl_zk select c.klient_id, b.CNUM, '', a.CLNTSN,a.mobil,'[BANKA].[dbo].[PROFILE_DATA]' from [BANKA].[dbo].[PROFILE_DATA] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where (Right(RTRIM(a.mobil),Len(@ptelc_e)) = @ptelc_e)
	--pevna linka
	insert into @tbl_zk select c.klient_id, b.CNUM, '', a.CLNTSN,a.telefon,'[BANKA].[dbo].[PROFILE_DATA]' from [BANKA].[dbo].[PROFILE_DATA] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where Right(RTrim(a.telefon),9) = @ptelc

	insert into @tbl_zk select c.klient_id, b.CNUM, '', a.CLNTSN,a.telefon,'[BANKA].[dbo].[PROFILE_DATA]' from [BANKA].[dbo].[PROFILE_DATA] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where (Right(RTRIM(a.telefon),Len(@ptelc_e)) = @ptelc_e)

	--IB ziadost o PP
	insert into @tbl_zk select distinct c.klient_id, b.CNUM, '', a.CLNTSN,a.telefon,'[BANKA].[dbo].[ZIADOST_PP_IN_EXTENSION]' from [BANKA].[dbo].[ZIADOST_PP_IN_EXTENSION] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where Right(RTrim(a.telefon),9) = @ptelc
	
	insert into @tbl_zk select distinct c.klient_id, b.CNUM, '', a.CLNTSN,a.telefon,'[BANKA].[dbo].[ZIADOST_PP_IN_EXTENSION]' from [BANKA].[dbo].[ZIADOST_PP_IN_EXTENSION] a 
	inner join Banka..ibsclnt b on a.clntsn collate Slovak_CI_AS = b.CLNTSN
	join MAC..klient c on b.CNUM collate Slovak_CI_AS = c.cnum 
	where (Right(RTRIM(a.telefon),Len(@ptelc_e)) = @ptelc_e)

	insert into @tZoznamKlientov select distinct * from @tbl_zk

	RETURN 
END

GO
```
